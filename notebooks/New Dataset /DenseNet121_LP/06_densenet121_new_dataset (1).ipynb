## Bước 4: Khởi tạo Kiến trúc DenseNet-121 (Pure From Scratch)
* Xây dựng kiến trúc DenseNet-121 với 4 khối Dense Block `(6, 12, 24, 16)` và Transition Layers thu nhỏ số kênh.
* Kết nối tắt dày đặc (Dense Connectivity): Đầu ra của mỗi lớp được ghép nối (concatenation) với tất cả các lớp phía trước, giúp tái sử dụng đặc trưng và giảm suy hao gradient hiệu quả.
* Khởi tạo toàn bộ ~6.99M tham số ngẫu nhiên theo phân phối Kaiming He Normal, không sử dụng bất kỳ trọng số có sẵn nào.

```python
class _DenseLayer(nn.Module):
    def __init__(self, in_features, growth_rate, bn_size=4):
        super(_DenseLayer, self).__init__()
        self.bn1 = nn.BatchNorm2d(in_features)
        self.conv1 = nn.Conv2d(in_features, bn_size * growth_rate, kernel_size=1, bias=False)
        self.bn2 = nn.BatchNorm2d(bn_size * growth_rate)
        self.conv2 = nn.Conv2d(bn_size * growth_rate, growth_rate, kernel_size=3, padding=1, bias=False)

    def forward(self, x):
        out = self.conv1(torch.relu(self.bn1(x)))
        out = self.conv2(torch.relu(self.bn2(out)))
        return torch.cat([x, out], 1)

class _DenseBlock(nn.Module):
    def __init__(self, num_layers, in_features, growth_rate, bn_size=4):
        super(_DenseBlock, self).__init__()
        layers = []
        for i in range(num_layers):
            layers.append(_DenseLayer(in_features + i * growth_rate, growth_rate, bn_size))
        self.block = nn.Sequential(*layers)

    def forward(self, x):
        return self.block(x)

class _Transition(nn.Module):
    def __init__(self, in_features, out_features):
        super(_Transition, self).__init__()
        self.bn = nn.BatchNorm2d(in_features)
        self.conv = nn.Conv2d(in_features, out_features, kernel_size=1, bias=False)
        self.pool = nn.AvgPool2d(kernel_size=2, stride=2)

    def forward(self, x):
        out = self.conv(torch.relu(self.bn(x)))
        return self.pool(out)

class CustomDenseNet121(nn.Module):
    def __init__(self, growth_rate=32, block_config=(6, 12, 24, 16), num_init_features=64, num_classes=37):
        super(CustomDenseNet121, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, num_init_features, kernel_size=7, stride=2, padding=3, bias=False),
            nn.BatchNorm2d(num_init_features),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=3, stride=2, padding=1)
        )
        num_features = num_init_features
        for i, num_layers in enumerate(block_config):
            block = _DenseBlock(num_layers=num_layers, in_features=num_features, growth_rate=growth_rate)
            self.features.add_module(f'denseblock{i + 1}', block)
            num_features = num_features + num_layers * growth_rate
            if i != len(block_config) - 1:
                trans = _Transition(in_features=num_features, out_features=num_features // 2)
                self.features.add_module(f'transition{i + 1}', trans)
                num_features = num_features // 2

        self.features.add_module('norm5', nn.BatchNorm2d(num_features))
        self.classifier = nn.Linear(num_features, num_classes)
        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.normal_(m.weight, 0, 0.01)
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)

    def forward(self, x):
        features = self.features(x)
        out = torch.relu(features)
        out = nn.functional.adaptive_avg_pool2d(out, (1, 1))
        out = torch.flatten(out, 1)
        return self.classifier(out)

model = CustomDenseNet121(num_classes=num_classes).to(device)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Đã chọn mô hình: Custom DenseNet-121 (From Scratch)")
print(f"Khởi tạo trọng số Kaiming He Normal hoàn tất!")
print(f"Tổng số tham số: {total_params:,}")
```

* Điều chỉnh tốc độ học theo chu kỳ Cosine Annealing giảm dần về 0.
* Tự động lưu checkpoint sau mỗi epoch (`latest.pth`, `best.pth`) và tự động phục hồi (Resume) trạng thái nếu bị gián đoạn.

```python
os.makedirs(args.checkpoint_dir, exist_ok=True)

criterion = nn.CrossEntropyLoss(label_smoothing=args.label_smoothing)
optimizer = optim.AdamW(model.parameters(), lr=args.lr, weight_decay=args.weight_decay)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=args.epochs, eta_min=1e-6)

best_ckpt = os.path.join(args.checkpoint_dir, "densenet121_augmented_best.pth")

# Tự động tìm kiếm checkpoint nếu được upload qua Kaggle Dataset (/kaggle/input)
if not os.path.exists(best_ckpt):
    kaggle_input = Path("/kaggle/input")
    if kaggle_input.exists():
        found_ckpts = list(kaggle_input.rglob("densenet121_augmented_best.pth"))
        if not found_ckpts:
            found_ckpts = list(kaggle_input.rglob("*densenet121*.pth"))
        if found_ckpts:
            import shutil
            shutil.copy(str(found_ckpts[0]), best_ckpt)
            print(f">> Đã tự động nạp checkpoint từ Kaggle Dataset: {found_ckpts[0]} -> {best_ckpt}")

start_epoch = 1
best_acc = 0.0
train_losses, test_losses = [], []
train_accs, test_accs = [], []

if os.path.exists(best_ckpt):
    ckpt = torch.load(best_ckpt, map_location=device, weights_only=False)
    model.load_state_dict(ckpt["model_state_dict"])
    optimizer.load_state_dict(ckpt["optimizer_state_dict"])
    if "scheduler_state_dict" in ckpt:
        scheduler.load_state_dict(ckpt["scheduler_state_dict"])
    else:
        scheduler.last_epoch = ckpt["epoch"]
    start_epoch = ckpt["epoch"] + 1
    best_acc = ckpt["best_acc"]
    history = ckpt["history"]
    train_losses, test_losses = history["train_loss"], history["test_loss"]
    train_accs, test_accs = history["train_acc"], history["test_acc"]
    print(f"Đã phục hồi trạng thái từ Epoch {ckpt['epoch']} với Test Acc cao nhất: {best_acc:.2f}%")

print(f"Bắt đầu huấn luyện DenseNet-121 từ Epoch {start_epoch} đến Epoch {args.epochs}...")

for epoch in range(start_epoch, args.epochs + 1):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, targets, _ in trainloader:
        inputs, targets = inputs.to(device), targets.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, preds = outputs.max(1)
        correct += preds.eq(targets).sum().item()
        total += targets.size(0)

    scheduler.step()
    train_loss = running_loss / total
    train_acc = 100.0 * correct / total

    # Đánh giá trực tiếp trên Benchmark Test Chuẩn (3.669 ảnh)
    model.eval()
    val_loss, val_correct, val_total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, targets, _ in testloader:
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            val_loss += loss.item() * inputs.size(0)
            _, preds = outputs.max(1)
            val_correct += preds.eq(targets).sum().item()
            val_total += targets.size(0)

    test_loss = val_loss / val_total
    test_acc = 100.0 * val_correct / val_total

    train_losses.append(train_loss)
    test_losses.append(test_loss)
    train_accs.append(train_acc)
    test_accs.append(test_acc)

    is_best = (test_acc > best_acc)
    if is_best:
        best_acc = test_acc
        torch.save({
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "best_acc": best_acc,
            "history": {
                "train_loss": train_losses,
                "test_loss": test_losses,
                "train_acc": train_accs,
                "test_acc": test_accs
            }
        }, best_ckpt)

    print(f"Epoch {epoch:02d}/{args.epochs:02d} | Train Loss: {train_loss:.3f} | Train Acc: {train_acc:.2f}% | Test Loss: {test_loss:.3f} | Test Acc: {test_acc:.2f}%")

print(f"\nHuấn luyện hoàn tất! Kết quả Test Acc cao nhất đạt: {best_acc:.2f}%")
```
